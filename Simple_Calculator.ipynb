```python
# Simple Calculator

num1 = float(input("Enter the first number: "))
num2 = float(input("Enter the second number: "))

operation = input("Choose an operation (+, -, *, /): ")

if operation == "+":
    result = num1 + num2
    print(f"✅ Result: {num1} + {num2} = {result}")

elif operation == "-":
    result = num1 - num2
    print(f"✅ Result: {num1} - {num2} = {result}")

elif operation == "*":
    result = num1 * num2
    print(f"✅ Result: {num1} * {num2} = {result}")

elif operation == "/":
    if num2 == 0:
        print("❌ Error: You can't divide by zero!")
    else:
        result = num1 / num2
        print(f"✅ Result: {num1} / {num2} = {result}")

else:
    print("❌ Invalid operation. Please use +, -, *, or /")
```

```python
import math

# This list will store all calculations
history = []

while True:
    # ====== DISPLAY MENU ======
    print("\n========================")
    print("      SMART CALCULATOR  ")
    print("========================")
    print("1. Add")
    print("2. Subtract")
    print("3. Multiply")
    print("4. Divide")
    print("5. Power")
    print("6. Square Root")
    print("7. View History")
    print("8. Clear History")
    print("9. Exit")
    print("========================")

    # ====== GET MENU CHOICE ======
    choice = input("Enter your choice (1-9): ")

    # ====== OPTION 1: ADD ======
    if choice == "1":
        num1 = float(input("Enter first number: "))
        num2 = float(input("Enter second number: "))
        result = num1 + num2
        print(f"✅ Result: {num1} + {num2} = {result}")
        history.append(f"{num1} + {num2} = {result}")

    # ====== OPTION 2: SUBTRACT ======
    elif choice == "2":
        num1 = float(input("Enter first number: "))
        num2 = float(input("Enter second number: "))
        result = num1 - num2
        print(f"✅ Result: {num1} - {num2} = {result}")
        history.append(f"{num1} - {num2} = {result}")

    # ====== OPTION 3: MULTIPLY ======
    elif choice == "3":
        num1 = float(input("Enter first number: "))
        num2 = float(input("Enter second number: "))
        result = num1 * num2
        print(f"✅ Result: {num1} * {num2} = {result}")
        history.append(f"{num1} * {num2} = {result}")

    # ====== OPTION 4: DIVIDE ======
    elif choice == "4":
        num1 = float(input("Enter first number: "))
        num2 = float(input("Enter second number: "))
        if num2 == 0:
            print("❌ Error: You can't divide by zero!")
        else:
            result = num1 / num2
            print(f"✅ Result: {num1} / {num2} = {result}")
            history.append(f"{num1} / {num2} = {result}")

    # ====== OPTION 5: POWER ======
    elif choice == "5":
        num1 = float(input("Enter the base number: "))
        num2 = float(input("Enter the power/exponent: "))
        result = num1 ** num2
        print(f"✅ Result: {num1} ^ {num2} = {result}")
        history.append(f"{num1} ^ {num2} = {result}")

    # ====== OPTION 6: SQUARE ROOT ======
    elif choice == "6":
        num1 = float(input("Enter a number: "))
        if num1 < 0:
            print("❌ Error: Cannot find square root of a negative number!")
        else:
            result = math.sqrt(num1)
            print(f"✅ Result: √{num1} = {result}")
            history.append(f"√{num1} = {result}")

    # ====== OPTION 7: VIEW HISTORY ======
    elif choice == "7":
        print("\n📋 Calculation History:")
        print("------------------------")
        if len(history) == 0:
            print("No calculations yet!")
        else:
            for i, item in enumerate(history, 1):
                print(f"{i}. {item}")
        print("------------------------")

    # ====== OPTION 8: CLEAR HISTORY ======
    elif choice == "8":
        history.clear()
        print("🗑️ History cleared!")

    # ====== OPTION 9: EXIT ======
    elif choice == "9":
        print("👋 Goodbye! Thanks for using Smart Calculator!")
        break

    # ====== INVALID CHOICE ======
    else:
        print("❌ Invalid choice! Please enter a number between 1 and 9.")
```
